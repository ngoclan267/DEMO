```python
# read_social_listening_db.py
import os
import sqlite3
import pandas as pd

DB_PATH = "social_listening.db"

if not os.path.exists(DB_PATH):
    raise FileNotFoundError(f"{DB_PATH} not found in current directory")

conn = sqlite3.connect(DB_PATH)
cur = conn.cursor()

cur.execute("SELECT name FROM sqlite_master WHERE type='table' AND name NOT LIKE 'sqlite_%';")
tables = [r[0] for r in cur.fetchall()]
print("Tables:", tables)

for table in tables:
    print(f"\n--- Table: {table} ---")
    cur.execute(f"PRAGMA table_info({table})")
    schema = [(c[1], c[2]) for c in cur.fetchall()]
    print("Schema:", schema)
    df = pd.read_sql_query(f"SELECT * FROM {table} LIMIT 100", conn)
    print(df.head())

cur.close()
conn.close()
```

```python
from IPython.display import display

pd.set_option("display.max_colwidth", 120)
pd.set_option("display.max_rows", 40)
pd.set_option("display.width", 150)
pd.set_option("display.colheader_justify", "left")

readable_table = df.copy()
readable_table["sent_at"] = pd.to_datetime(readable_table["sent_at"])

display(readable_table)
```

```python
# Đọc tất cả các bảng trong DB_PATH vào dict các DataFrame và hiển thị
conn = sqlite3.connect(DB_PATH)
cur = conn.cursor()
cur.execute("SELECT name FROM sqlite_master WHERE type='table' AND name NOT LIKE 'sqlite_%';")
tables = [r[0] for r in cur.fetchall()]

tables_dfs = {}
for table in tables:
    tables_dfs[table] = pd.read_sql_query(f'SELECT * FROM "{table}"', conn)
    print(f"\n--- Table: {table} ({len(tables_dfs[table])} rows, {len(tables_dfs[table].columns)} cols) ---")
    display(tables_dfs[table].head(100))

cur.close()
conn.close()
```

```python
Các bảng hiện có trong DB này phục vụ các chức năng sau:

- `topics`: lưu thông tin các chủ đề / lĩnh vực đang theo dõi. Thường chứa id, người quản lý/owner, tên chủ đề, và các thông tin metadata khác.

- `posts`: lưu các bản ghi bài viết/đoạn hội thoại/mention thực tế thu thập được từ social listening. Nó liên kết với `topic_id`, chứa nội dung, nguồn kênh, thời gian, và các thuộc tính liên quan đến sentiment/feedback.

- `pain_points`: lưu các “điểm đau” / vấn đề cụ thể thuộc về một topic. Mỗi bản ghi gắn với `topic_id` và mô tả một vấn đề như lỗi đăng nhập, hiệu năng, bảo mật, v.v.

- `notifications`: lưu các thông báo cảnh báo/alert liên quan đến topic và pain point. Có các trường như `channel`, `title`, `message`, `sent_at`, và `read` để quản lý việc gửi và trạng thái đã đọc.

Nói ngắn gọn: `topics` là nhóm chủ đề; `posts` là dữ liệu thu thập; `pain_points` là các vấn đề cụ thể; `notifications` là cảnh báo dựa trên vấn đề đó.
```

```python
`pain_points` lưu mỗi bản ghi là một “điểm đau” / vấn đề cụ thể của một `topic`.

Thông thường:
- `id`: khóa chính của pain point
- `topic_id`: liên kết về `topics`
- `title`: tên / mô tả ngắn gọn vấn đề
- các trường khác có thể là mô tả chi tiết, loại vấn đề, metadata…

Nói cách khác: `pain_points` không phải là post, mà là bảng định nghĩa các vấn đề cụ thể (ví dụ: lỗi đăng nhập, hiệu năng, bảo mật) gắn với một chủ đề theo dõi.
```

```python
Vì người dùng chỉ đăng phản hồi đánh giá app nên `topic` và `pain_point` phải được suy ra từ nội dung đó, chứ không phải tự nhiên có sẵn.

Thông thường quy trình sẽ là:

- `topics` là những nhóm chủ đề bạn quan tâm theo dõi (ví dụ: trải nghiệm đăng nhập, hiệu năng, bảo mật, thanh toán…).
- `pain_points` là các vấn đề cụ thể trong từng topic (ví dụ: “lỗi đăng nhập”, “chậm khởi động”, “khoá tài khoản”, “phí ẩn”).
- Dữ liệu đánh giá trong `posts` / `notifications.message` chứa văn bản phản hồi.
- Một bước xử lý/nghiên cứu sẽ phân tích văn bản đó rồi gán `topic_id` và `pain_point_id`.

Cách gán có thể là:

- quy tắc từ khóa: nếu review chứa “đăng nhập”, “mật khẩu”, “auth” -> gán vào topic/pain point đăng nhập;
- phân loại văn bản NLP hoặc model machine learning;
- đánh dấu thủ công / annotation bởi analyst;
- kết hợp với sentiment để biết đây là phản hồi tiêu cực và cần tạo alert.

Vì vậy, nếu chỉ nhìn vào feedback text, bạn phải dùng thuật toán hoặc quy tắc để xác định:
- người dùng đang nói về lĩnh vực gì (`topic`),
- vấn đề cụ thể nào đang xảy ra (`pain_point`).

Trong DB hiện tại, bảng `notifications` đã lưu kết quả gán đó: mỗi bản ghi có `topic_id` và `pain_point_id`, nghĩa là việc xác định đã được thực hiện trước đó.
```
